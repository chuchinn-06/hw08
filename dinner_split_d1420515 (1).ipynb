```python
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 100
budget_per_person = 300

# 1. 整理聚餐名稱：去除頭尾空白並轉成小寫
cleaned_event_name = event_name.strip().lower()

# 2. 整理餐費與人數並轉換型別
# 移除 "NT$", ",", 與空白，並轉成浮點數以便計算小數點
cleaned_meal_str = meal_amount.replace("NT$", "").replace(",", "").strip()
original_meal_amount = float(cleaned_meal_str)

# 將文字人數轉成整數
total_people = int(people)

# 3. 分步計算費用
service_fee = original_meal_amount * service_rate
total_amount = original_meal_amount + service_fee - discount
cost_per_person = total_amount / total_people

# 4. 判斷是否超過預算（回傳布林值）
is_over_budget = cost_per_person > budget_per_person

# 5. 輸出費用摘要
print(f"聚餐名稱：{cleaned_event_name}")
print(f"原始餐費：{original_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{total_people} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```

## 3. 問答題

### 問題一：為什麼 "4" 要先轉成數字，才能用於本題的分攤計算？
**回答**：
因為 `"4"` 在 Python 中被雙引號包圍，屬於字串型別（`str`）。在 Python 裡，字串與數字無法直接進行數學除法運算（例如執行 `1220 / "4"` 會引發 `TypeError` 錯誤）。因此，必須先透過 `int("4")` 將其轉換為整數（`int`），電腦才能正確理解並執行分攤金額的除法計算。

### 問題二：如果先扣折價券，再依剩餘餐費計算服務費，第一組的應付總額會變成多少？為什麼和題目規定的結果不同？
**回答**：
- **計算結果**：
  1. 先扣折價券：1,200 － 100 ＝ 1,100 元
  2. 再計算服務費（10%）：1,100 × 0.10 ＝ 110 元
  3. 應付總額：1,100 ＋ 110 ＝ 1,210 元（與規定的 1,220 元不同）。
- **原因**：
  餐廳服務費的收取標準一般是以「原始消費金額」作為計算基準，而非扣除優惠折價後的金額。若先折價再算服務費，服務費的計算基礎會變小，導致餐廳收到的服務費減少，這不符合一般餐飲業的計費規則與本題規定的商業邏輯。
